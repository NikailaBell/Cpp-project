# **Methods**

# Exploring average minimum temperature

```r
# Load the data
slc_data <- read.csv("SLCyearly.csv")
model <- lm(tmin ~ Year, data = slc_data)
plot(slc_data$Year, slc_data$tmin,
     main = "Average Minimum Temperature vs Year",
     xlab = "Year", ylab = "Average Minimum Temperature",
     pch = 19, col = "blue")

abline(model, col = "red", lwd = 2)
```

**Regression Model Equation**:

tmin = β0​+β1​(Year)

 tmin= -124.86 + .0836(year)

 the average change in min temperature per year is .084 degrees F

β0 =intercept

β1 =change in average minimum temperature per year

 **What would a slope of zero imply about the relationship between average minimum temperature and year?**

𝛽1 = 0  no change in temperature over time

𝛽1 > 0  average minimum temperature is increasing each year

𝛽1 < 0 average minimum temperature is decreasing each year

**Is it possible that average minimum temperature and year are not related?**

H0:β1 = 0(no relationship)

H0:β1 ≠ 0(there is a relationship)


```r
model <- lm(tmin ~ Year, data = slc_data)
summary(model)
```

We Reject HO. There is a strong, statistically significant relationship  minimum temperatures are increasing over time. Given our small P-value 7.19e-16.

 looking at our R-Squared value we can see that 62.4% of the variation in average minimum temperature is explained by year.

**Assumptions of linear regression with no transformations**:

**Linearity:**

```r
plot(model$fitted.values, model$residuals,
     xlab = "Fitted values",
     ylab = "Residuals",
     main = "Residuals vs Fitted")
abline(h = 0, col = "red")
```

The residuals versus fitted values plot shows no clear patterns or systematic curvature, indicating that the relationship between year and average minimum temperature is adequately modeled as linear.

**Normality of residuals:**

```r
qqnorm(model$residuals)
qqline(model$residuals, col = "red")
shapiro.test(residuals(model))
```

The normal Q-Q plot of the residuals shows that most points closely follow the reference line, indicating that the residuals are approximately normally distributed, with only slight deviations at the ends. We also included a Shapiro-Wilk normality test. Since the P-value is 0.8461 we fail to reject the null hypothesis that the residuals are normally distributed.

**Constant Variance**:

```r
install.packages('lmtest')
library(lmtest)

# Perform the test
bptest(model)

# Scale-Location Plot
plot(model$fitted.values, sqrt(abs(residuals(model))),
     xlab = "Fitted Values", ylab = "Sqrt(|Residuals|)",
     main = "Scale-Location Plot (Original tmin)")
```

The scale-location plot shows a fairly random spread of the square root of the absolute residuals across the fitted values, indicating no clear pattern of increasing or decreasing variance. Consistently, the Breusch-Pagan test yielded a p-value of 0.93, providing strong evidence that the assumption of constant variance (homoscedasticity) is satisfied for this model.

**Indenpendence**:

```r
install.packages("lmtest")
library(lmtest)

dwtest(model)
```

The Durbin-Watson test yielded a statistic of 1.53 with a p-value of 0.017, indicating significant positive autocorrelation in the residuals. This suggests that the independence assumption of the regression model may be violated, as residuals from one year are correlated with those from the previous year.

# Exploring average minimum temperature with a log transformation

```r
# Load the data
slc_data <- read.csv("SLCyearly.csv")

# Perform the log transformation on average minimum temperature
slc_data$log_tmin <- log(slc_data$tmin)

# View the first few rows to confirm
head(slc_data)
```

```r
# Fit the log-transformed model
log_model <- lm(log_tmin ~ Year, data = slc_data)

# Create the plot
plot(slc_data$Year, slc_data$log_tmin,
     main = "Log of Average Min Temperature vs Year",
     xlab = "Year", ylab = "Log(Average Min Temp)",
     pch = 19, col = "blue")
abline(log_model, col = "red", lwd = 2)
```

**Interpretation of the model**:

log(tmin) = β0 + β1​(year)

log(tmin)= -0.3253 + .002(year)

because the slope is logged it is interpreted as a percentage change, so the average min temperature increases by .20% for each year increase.

In a log-linear model (where only the output y is logged), the relationship between the year (x) and temperature (y) changes from an additive effect to a multiplicative effect.

one-year increase in Year is associated with an approximate 100 * β1 increase in average minimum temperature. Because of the log transformation, this corresponds to a percent change in actual temperature.

**Transformed regression model**:

To interpret results in the original temperature scale, the model is exponentiated:

tmin=e ^ (β0​+β1​(Year))

**Transformed vs linear model scatter plot:**

```r
slc <- read.csv("SLCyearly.csv")
model_linear <- lm(tmin ~ Year, data = slc)
plot(tmin ~ Year, data = slc,
     main = "Comparison of Linear and Log Models",
     xlab = "Year", ylab = "Average Minimum Temperature")

abline(model_linear, col = "red", lwd = 2)

# Add transformed model curve
curve(exp(coef(log_model)[1] + coef(log_model)[2]*x),
      add = TRUE, col = "blue", lwd = 2)
```

Both the linear and log-transformed models show a similar increasing trend in average minimum temperature over time. The near overlap of the two models suggests that the relationship between Year and temperature is approximately linear, with no strong evidence of exponential growth within the observed time period.

**H₀: 𝛽1=0 (no relationship)**

**H₁: 𝛽1≠0 (relationship exists)**



```r
summary(log_model)
```

A t-test for the slope yields a p-value of 8.87×10⁻¹⁶, so **we reject the null hypothesis** and conclude there is strong evidence that average minimum temperature and year are related.The log-transformed regression model indicates that average minimum temperature increases at an approximate rate of 0.2035% per year. The model explains about 62.16% of the variability in the data, suggesting a moderately strong fit.

**Assumptions of linear regression for our log transformed model:**

**Linearity**:

```r
plot(log_model$fitted.values, residuals(log_model),
     xlab = "Fitted Values", ylab = "Residuals",
     main = "Residuals vs Fitted")
abline(h = 0, col = "red")
```

The residuals are generally scattered around zero with no strong systematic pattern.The linearity assumption is reasonably satisfied, although there may be very slight nonlinearity.



**Normality of residuals**:

```r
qqnorm(residuals(log_model))
qqline(residuals(log_model), col = "red")
shapiro.test(residuals(log_model))
```

The residuals are approximately normally distributed, with minor deviations at the extremes that are not severe. The P-Value in the Shaprio test is greater than .05 so we fail to reject normality.

**Constant Variance**:

```r
plot(log_model$fitted.values, sqrt(abs(residuals(log_model))),
     xlab = "Fitted Values", ylab = "Sqrt(|Residuals|)",
     main = "Scale-Location Plot")

install.packages("lmtest")
library(lmtest)

bptest(log_model)
```

Since the p-value (0.6532) is greater than 0.05, we fail to reject our null hypothesis.
There is no statistical evidence of heteroscedasticity, meaning the assumption of constant variance is satisfied.
The plot suggests that the variance of the residuals is approximately constant, supporting the homoscedasticity assumption.

**Independence**:

```r
install.packages("lmtest")
library(lmtest)

dwtest(log_model)
```

The independence assumption was tested using the Durbin–Watson test (DW = 1.5355, p = 0.01825). Since the p-value is less than 0.05, there is evidence of positive autocorrelation in the residuals, indicating that the independence assumption is violated.

# Comparing average minimum temperatures via ANOVA

**Divide your data set into 3 equal groups and report the date ranges for each group**

Group 1: 1949–1971 (first 23 years)

Group 2: 1972–1994 (next 23 years)

Group 3: 1995–2017 (last 23 years)





**Box plot of the data**

```r
# Create a grouping variable
slc_data$period <- cut(slc_data$Year,
                       breaks = c(1948, 1971, 1994, 2017),
                       labels = c("1949-1971", "1972-1994", "1995-2017"))

boxplot(tmin ~ period, data = slc_data,
        main = "Average Minimum Temperatures by Period",
        xlab = "Period",
        ylab = "Average Minimum Temperature",
        col = c("lightblue", "lightgreen", "salmon"))
```

**Anova Test**

H₀: The mean minimum temperatures are the same across all 3 periods

H₁: At least one period has a different mean minimum temperature


```r
anova_model <- aov(tmin ~ period, data = slc_data)
summary(anova_model)
```

The F Value is 52.5 this is very high, indicating that the variation between the periods is much greater than the variation within the periods.

The P-Value is extremely small (much smaller than 0.05), so we **reject the null** **hypothesis** (H₀). There is a statistically significant difference in average minimum temperatures among the three periods. Temperatures have clearly increased over time.

**Post Hoc Test**

```r
TukeyHSD(anova_model)
```

Since all P-Values are small. This means all three periods are significantly different from each other in terms of average minimum temperature.

1972–1994 vs 1949–1971: Average minimum temperature increased by ~2.23°F.

1995–2017 vs 1949–1971: Average minimum temperature increased by ~4.04°F.

1995–2017 vs 1972–1994: Average minimum temperature increased by ~1.81°F.

**Assumptions of ANOVA**

**Normality of residuals**:

```r
# Shapiro-Wilk test
shapiro.test(residuals(anova_model))

# Q-Q plot
qqnorm(residuals(anova_model))
qqline(residuals(anova_model), col = "red")
```

The Q–Q plot indicates that the data are approximately normally distributed in the center, but show noticeable deviations in the tails, suggesting heavier tails and slight positive skewness. The Shapiro-Wilk normality test tell us that our assumption of ANOVA is satisfied. Since our P-Value is greater than 0.05.

**Homogeneity of variance (Homoscedasticity)**

```r
install.packages('car')
library(car)
leveneTest(tmin ~ period, data = slc_data)
```

A Levene’s test was conducted to assess the equality of variances in average minimum temperatures across the three time periods. The test indicated no significant difference in group variances (F = 0.602, p = 0.551), suggesting that the assumption of homogeneity of variance required for ANOVA is satisfied.

## Nights Warming Faster Than Days: Analysis and Explanation

Our regression analysis shows that average minimum temperatures in Salt Lake City are increasing much faster than average maximum temperatures. Average minimum temperature is a measure of nighttime temperature because the daily low typically happens in the early morning hours before sunrise, when there is no solar heating and the atmosphere has been releasing heat all night. Average maximum temperature is a measure of daytime temperature because the daily high usually occurs in the early afternoon when the sun is strongest.

The linear regression model for average minimum temperature gives a slope of 0.0836 degrees F per year (about 0.84 degrees per decade), with a p-value of 7.19 x 10^-16 and an R-squared of 0.624. Average maximum temperature only increases at 0.021 degrees F per year, with an R-squared of 0.079. Over the 68 years of data (1949 to 2017), average minimum temperature went up by about 5.7 degrees F while average maximum temperature only went up by about 1.5 degrees F. That means nighttime temperatures in SLC are warming roughly 3.9 times faster than daytime temperatures.

This pattern matches what researchers have found at the national level. Pierre-Louis and Popovich (2018) report that across the United States, overnight low temperatures have increased at 1.4 degrees F per century since 1895, compared to only 0.7 degrees F per century for daytime highs. So nationally, nights are warming about twice as fast as days. Our SLC data shows an even bigger gap, which could be explained by local factors like the Salt Lake Valley being enclosed by mountains that trap heat, as well as SLC's rapid growth and urbanization over this time period. Pierre-Louis and Popovich give two main explanations for why nights are warming faster than days. The first is increased moisture in the atmosphere, which holds in heat at night and prevents temperatures from dropping as much as they used to. The second is the urban heat island effect, where cities absorb heat in pavement and buildings during the day and slowly release it at night. The article points out that cities can be up to 22 degrees F warmer than surrounding less developed areas because of all their heat-absorbing surfaces and lack of trees and green spaces. Both of these factors are very relevant to Salt Lake City.


## Green Spaces, Disproportionate Impact, and Social Equity

Pierre-Louis and Popovich (2018) explain that green spaces like parks, trees, and vegetation help cool neighborhoods through shade and a process called evapotranspiration, where plants release water vapor that cools the surrounding air. Neighborhoods with fewer green spaces have more pavement and buildings that absorb and hold heat, which makes the nighttime warming problem worse.

This issue does not affect everyone equally. In cities like Los Angeles, Asian-American, Black, and Hispanic residents are more likely to live in the hotter parts of the city compared to white residents. Pierre-Louis and Popovich (2018) explain that this is due to a combination of factors including less access to green spaces, differences in elevation, and proximity to the coast. This means communities of color face a higher risk from the health effects of warming nights. The article also points out that older people, children, and people who are already sick are especially vulnerable to heat, along with outdoor workers in agriculture and construction who cannot simply move their work indoors. When nighttime temperatures stay high, the body does not get a chance to cool down and recover from the heat of the day. Lara Cushing, a professor of environmental epidemiology at San Francisco State University, is quoted in the article saying that the combination of high daytime and high nighttime temperatures can be really lethal for this reason.


## Rates of Night vs. Day Warming: SLC Data vs. National Findings

The national finding from Pierre-Louis and Popovich (2018) that nights are warming twice as fast as days is already a significant result. Our SLC data shows an even larger difference. Average minimum temperature is rising at 0.0836 degrees F per year while average maximum temperature is only rising at 0.0214 degrees F per year, giving a ratio of about **3.9 to 1**. Salt Lake City's rate of nighttime warming appears to be more severe than the national average. This could be because of the valley's geography, since the mountains surrounding SLC trap warm air in the basin. It could also relate to the shrinking Great Salt Lake, which historically helped moderate local temperatures, as well as decreased snowpack in the mountains and the significant population and urban growth the region has seen over the past several decades. Any of these factors could be pushing SLC's nighttime warming rate above the national trend.

## Shifting Temperature Distributions and the ANOVA Box Plot

Popovich and Pearce (2017) use a series of histograms based on data from NASA climate scientist James Hansen to show how summer temperatures in the Northern Hemisphere have shifted over time. During the baseline period of 1951 to 1980, summer temperatures were roughly evenly split into three categories: about one third were considered cold, one third were normal, and one third were hot. By 2005 to 2015, that distribution had changed significantly. About two thirds of values were now in the hot category, and nearly 15 percent fell into a brand new category called extremely hot that barely existed during the baseline period. The histograms are useful because they show it is not just the average temperature going up but the whole distribution shifting to the right, with far more extreme heat events and far fewer cool summers than before. Climate scientists quoted in the article note that even small increases in the average temperature can have big effects on the extremes, which is what makes this trend so concerning.

Our ANOVA box plot for SLC average minimum temperature across the three time periods shows a very similar pattern. The median minimum temperature goes up in each successive group, with the boxes rising from the 1949 to 1971 period through to 1995 to 2017. The Tukey post hoc test confirmed that all three periods are significantly different from each other. Average minimum temperature increased by about 2.23 degrees F from the first period to the second, by about 1.81 degrees F from the second to the third, and by about 4.04 degrees F from the earliest to the most recent period. This means the warming has been steady and progressive across the full record rather than one big jump at a single point in time, which mirrors the gradual rightward shift of the histograms described by Popovich and Pearce.

# **Conclusion**

## Summary of Statistical Findings

Our analysis of Salt Lake City temperature data from 1949 to 2017 found strong statistical evidence that temperatures have been increasing over time, especially at night. The linear regression of average minimum temperature on year gave a slope of 0.0836 degrees F per year (R-squared = 0.624, p-value = 7.19 x 10^-16), meaning that year is a strong and significant predictor of nighttime temperature and that about 62.4% of the variation in average minimum temperature is explained by time. The log-transformed regression model gave nearly the same results (R-squared = 0.622, p-value = 8.87 x 10^-16). Both models passed the key regression assumptions of linearity, normality of residuals, and constant variance. The independence assumption was technically violated in both models based on the Durbin-Watson test (p = 0.017), which is a common issue with time series data. We chose to keep the linear model because it was easier to interpret and had a slightly higher R-squared.

The ANOVA test comparing the three time periods (1949 to 1971, 1972 to 1994, and 1995 to 2017) came back highly significant (F = 52.5), confirming that average minimum temperatures are meaningfully different across eras. The Tukey post hoc test showed that all three groups were significantly different from each other. Average minimum temperature increased by about 2.23 degrees F from the first to the second period, by about 1.81 degrees F from the second to the third, and by about 4.04 degrees F from the earliest to the most recent period. The ANOVA assumptions were also satisfied: the Shapiro-Wilk test confirmed normality of residuals (p > 0.05), and the Levene test confirmed equal variances across the three groups (F = 0.602, p = 0.551). Average maximum temperature showed a much weaker trend (slope = 0.021 degrees F per year, R-squared = 0.079, p = 0.020), which confirms that the warming in SLC has been much stronger for nighttime lows than for daytime highs.

## The Role of Statistics in Society and Sustainability

Statistics is important because it gives us a way to take large amounts of raw data and figure out whether patterns we see are real or just due to random chance. Without tools like regression and ANOVA, we would not be able to tell if the temperature changes we see in the SLC data are part of a true long-term trend or just normal year-to-year variation. Being able to make that distinction is really important for making good decisions about environmental policy and sustainability.

That said, statistics also has limits. Since our data comes from an observational study and not a controlled experiment, we cannot say that time itself is causing temperatures to rise. We can only say that there is a strong and significant association between year and temperature. There are also potential confounding variables we cannot fully account for, like the urban heat island effect, the fact that the weather station was relocated at some point and different instruments were used over time, and natural climate cycles like ENSO that can cause multi-year swings in temperature. Statistics helps us see what is happening in the data, but it cannot tell us exactly why it is happening or prove causation on its own.

Even with those limitations, the evidence from this project is useful and meaningful. Results like these can help city planners and policymakers make decisions about things like expanding green spaces, updating building standards for heat resilience, and planning cooling resources for vulnerable populations, even without having a perfectly controlled experiment. This project also connects to UNESCO Sustainable Development Goal 13, which focuses on taking action on climate change. Our finding that nighttime temperatures in SLC have risen by nearly 5.7 degrees F over 68 years is exactly the kind of local, data-driven evidence that supports the need for action at both the city and global level.
